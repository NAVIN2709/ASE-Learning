MACE Calculator

Mace model we should have in the same root , load it thorugh this MACECalculator and use it to predict energy and forces

```python
#Using MACE Calculator

from mace.calculators import MACECalculator , mace_mp
# import inspect
# print(inspect.signature(MACECalculator)) read all inputs
"""
 for a model which we have trained and we want to use
"""
# calc = MACECalculator(
#     model_path="", #Path to the .model
#     device="cuda"
# )
"""
mace_mp is a pretrained mace model . We downloading 'https://github.com/ACEsuit/mace-mp/releases/download/mace_mp_0/2023-12-03-mace-128-L1_epoch-199.model'
"""
calc = mace_mp(
    model="medium",
    device="cuda",
)
print(calc)

atoms = read("structure.cif")
atoms.calc = calc

energy = atoms.get_potential_energy()
print("Energy=",energy)
print("Energy per atom=",energy/len(atoms))

forces = atoms.get_forces()
print("Forces per atom=",forces) # Forces are in eV/Å
print("Max Force=",forces.max())
print("Min Force=",forces.min())

"""
[ 2.75956700e-06  6.11048651e+00  1.10867404e-06] => [Fx,Fy,Fz]
"""
```

Now lets go to MACE-ASE Optimizer and geometry relaxation

```python
from ase.optimize import BFGS #BFGS is an optimization algorithm.
from mace.calculators import mace_mp
relaxed = atoms.copy()

calc = mace_mp(
    model="medium",
    device="cuda",
)

relaxed.calc = calc

print("Initial energy:",
      relaxed.get_potential_energy())

print("Initial max force:",
      abs(relaxed.get_forces()).max())

opt = BFGS(relaxed) # Create the optimizer

opt.run(fmax=0.05) #Keep optimizing until the largest force is below 0.05 eV/Å. This runs the full loop and prints each step time energy and fmax , here forces are in eV/Å

print("Final energy:",
      relaxed.get_potential_energy())

print("Final max force:",
      abs(relaxed.get_forces()).max())
```

```python
print(relaxed)
print(atoms)

import numpy as np

before = atoms.get_positions()
after = relaxed.get_positions()

displacements = after - before
movement = np.linalg.norm(after - before, axis=1)
print("Atom    Element    Displacement (Å)")

for i, atom in enumerate(atoms): #Prints the displacement from the original dataset to the relaxed positions
    print(
        i,
        atom.symbol,
        displacements[i]
    )
    print(f"{i:2d}  {atom.symbol:2s}  moved = {movement[i]:.4f} Å") #Shows how much the atoms moved (eg:0  V   moved = 0.4682 Å)

largest = np.argsort(movement)[::-1]

for i in largest[:10]: #Largest moved atoms
    print(
        f"Atom {i:2d} ({atoms[i].symbol}): "
        f"{movement[i]:.4f} Å"
    )
```

Now this is how we would arrange data in the .extxyz format using ase to train the MACE to predict energy and forces

              DFT
               │
               │ reference
               ↓
        ┌──────────────┐
        │    Dataset   │
        │   extxyz     │
        └──────┬───────┘
               ↓
             MACE
               ↓
       Neural network
               ↓
      predicted Energy
      predicted Forces

MACE adjusts its parameters to ensure MACE prediction ≈ DFT result

For Energy
DFT energy       = -272.31 eV
MACE prediction  = -272.28 eV ( small error )

For Forces
DFT force        = [0.12, -0.03, 0.05]
MACE force       = [0.11, -0.02, 0.06]

Now the actual ML part starts

We actually split the data

                  ALL DATA
                     │
          ┌──────────┼──────────┐
          ↓          ↓          ↓
       TRAIN        VALID       TEST
        ~80%         ~10%        ~10%
          │           │           │
          ↓           ↓           ↓
       Learn       Monitor      Final
       model       training     evaluation

Training = homework
Validation = practice exam
Test = final exam

Structure
   ↓
MACE prediction
   ↓
Compare with DFT
   ↓
Calculate loss
   ↓
Update neural-network parameters

Adjacent frames can be extremely familiar , data leakage might happen , so we enforce difference splitting techniques like
TRAIN
relaxed + distorted + thermal
VALIDATION
some held-out configurations
TEST
different configurations / structures

```python
from ase.io import write
data = read("relax.extxyz", index=":")
print(len(data))

import random
random.seed(42)
indices = list(range(len(data)))
random.shuffle(indices)
n = len(data)
n_train = int(0.8 * n)
n_valid = int(0.1 * n)
train = [data[i] for i in indices[:n_train]] # Training data
valid = [data[i] for i in indices[n_train:n_train+n_valid]] #Validation data
test = [data[i] for i in indices[n_train+n_valid:]] #Testing data

print("Train:", len(train))
print("Valid:", len(valid))
print("Test:", len(test))

write("train.extxyz", train, format="extxyz")
write("valid.extxyz", valid, format="extxyz")
write("test.extxyz", test, format="extxyz")
```
